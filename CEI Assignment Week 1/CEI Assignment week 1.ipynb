##Handling Missing Values

```python
df['discount'].fillna(df['discount'].median(),inplace=True)
```

```python
df['seller_name'].fillna(df['seller_name'].mode()[0],inplace=True)
```

```python
df['seller_information'].fillna(df['seller_information'].mode()[0],inplace=True)
```

```python
df.drop(columns=['videos'],inplace=True)
```

```python
df.drop(columns=['what_customers_said'],inplace=True)
```

```python
df.drop(columns=['variations'],inplace=True)
```

```python
df.isnull().sum()
```

##Basic Operations

```python
#selecting important colummns
df[['title','rating','discount','final_price','delivery_options']].head()
```

```python
#filtering for high rating
high_rating= df[df['rating']>4.5]
high_rating.head()
```

```python
#for high discount
high_discount = df[df['discount']>65]
high_discount.head()
```

```python
#for intial price higher(premium products)
premium_products=df[df['initial_price']>7000]
premium_products.head()
```

```python
#combining both rating and discount for the best deals
best_deals=df[(df['rating']>4.5) & (df['discount']>65)]
best_deals.head()
```

```python
print("highly rated products:",len(high_rating))
print("highly discounted products:",len(high_discount))
print("premium products:",len(premium_products))
print("best deals:",len(best_deals))
```

##Removing duplicates

```python
#checking whether duplicates exist or not
duplicate_count=df.duplicated().sum()
print("Duplicates before removing:",duplicate_count)
```

```python
#duplicates doesnt exist in dataset 
#if a dataset has any duplicates then we should do
df=df.drop_duplicates()
```

##Derived Column creation


**Creating a column for  total amount**

since the dataset doesn't contain quantity column , so total_amount is calculated using the initial price and discount 

formula : total amount = initial pricce - (initial price * discount / 100)

```python
df['total_amount']=df['initial_price']-(df['initial_price']*df['discount']/100)
print(df['total_amount'].head())
```
